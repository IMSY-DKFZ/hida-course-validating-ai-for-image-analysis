```python
def show_masks(reference, prediction1, prediction2):
    fig, axes = plt.subplots(1, 3, figsize=(9, 3))

    axes[0].imshow(reference)
    axes[0].set_title("Reference")
    axes[0].axis("off")

    axes[1].imshow(prediction1)
    axes[1].set_title("Prediction 1")
    axes[1].axis("off")

    axes[2].imshow(prediction2)
    axes[2].set_title("Prediction 2")
    axes[2].axis("off")

    plt.show()
```
